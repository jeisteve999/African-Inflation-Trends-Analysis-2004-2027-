# Descriptive Statistics Interpretation

## The dataset contains 1,531 valid positive inflation observations covering multiple countries and regional economic groups between 2004 and 2027.

# Findings

# The average inflation rate across all observations is 10.48%, while the median inflation rate is 6.56%.
#  Since the mean is substantially higher than the median, the distribution is right-skewed, indicating that a small number of extreme inflation episodes significantly increase the overall average.
# Half of all observations fall below 6.56%, suggesting that moderate inflation is the most common scenario across the dataset.
#  The third quartile (75th percentile) is 10.68%, meaning that 75% of all observations report inflation rates below approximately 11%.
# The maximum observed inflation rate is 736.11%, highlighting the presence of severe hyperinflation episodes in specific countries.
# The standard deviation (32.84) is considerably larger than the median, reflecting substantial variability in inflation rates across countries and years.

# Inflation is generally moderate for most observations

## Although some countries experienced extremely high inflation, most observations are concentrated in relatively low inflation ranges.

## 50% of observations are below 6.56%
## 75% of observations are below 10.68%

## This indicates that extreme inflation events are uncommon but highly influential.

# These measures contains significant outliers

##The gap between:

##Median = 6.56%
## Maximum = 736.11%

## It is important to analyze. It reveals the existence of extraordinary inflationary crises.

## Countries such as Zimbabwe and South Sudan dominate the upper tail of the distribution and strongly influence aggregate statistics.

# Inflation experiences vary substantially across countries

##The high standard deviation suggests large differences in economic stability among countries.

## While some countries maintained relatively stable price levels, others experienced severe inflationary shocks.

# Hyperinflation events are concentrated in a small number of countries

## Preliminary analysis of the highest inflation observations shows that Zimbabwe and South Sudan account for many of the most extreme inflation values in the dataset.

## This suggests that regional and country-specific factors play a major role in inflation dynamics

```python
# Average Inflation by Country

country_avg= (imf_clean.groupby('Country')['Inflation'].mean().sort_values(ascending = False))

country_avg.head(15)
```

```python
#Visualization

import matplotlib.pyplot as plt

top15 = country_avg.head(15)

ax = top15.sort_values().plot(kind='barh', figsize=(12,6))
# title
plt.title('Top 15 Countries by Average Inflation')
# Not names in axis
plt.xlabel('Average Inflation (%')
plt.ylabel('')

for container in ax.containers:
    ax.bar_label(container,fmt='%.1f', padding=3)

# Remove the top and right spines

ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)
plt.xticks([])

plt.show()
```

```python
#  Inflation Volatility by Country

volatility = (imf_clean.groupby('Country')['Inflation'].std().sort_values(ascending = False))

top_vol = volatility.head(15)

top_vol
```

```python
# Visualization

ax= top_vol.sort_values().plot(kind='bar', figsize=(12,6), color = 'Red')

plt.title('Top 15 Countries by Inflation Volatility')
# Not names in axis
plt.xlabel('Inflation Volatility (%)')
plt.ylabel('')

for container in ax.containers:
    ax.bar_label(container,fmt='%.1f', padding=3)


ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)
plt.xticks(rotation= 45, ha= 'right')
plt.yticks([])
ax.invert_xaxis()

plt.tight_layout()
plt.show()
```

```python
# Inflation Trend Over Time

yearly_avg = (imf_clean.groupby('year')['Inflation'].mean())
yearly_avg
```

```python
# Visualization

plt.figure(figsize =(12,6))

plt.plot(yearly_avg.index,yearly_avg.values, marker='o')

plt.title('Global Average Inflation Trend')
plt.xlabel('Year')
plt.ylabel('Average Inflation (%)')

plt.grid(False)
plt.show()
```

```python
african_countries = [
    'Nigeria',
    'Kenya',
    'South Africa',
    'Egypt',
    'Morocco',
    'Zimbabwe'
]

africa = imf_clean[
    imf_clean['Country'].isin(african_countries)
]
```

```python
# Visualization

plt.figure(figsize=(12,6))

for country in african_countries:

    data = africa[
        africa['Country'] == country
    ]

    plt.plot(
        data['year'],
        data['Inflation'],
        marker='o',
        label=country
    )



plt.title('Inflation Trends Across Selected African Countries')
plt.xlabel('Year')
plt.ylabel('Inflation (%)')


ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)

plt.legend()
plt.show()
```
